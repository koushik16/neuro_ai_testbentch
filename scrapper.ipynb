```python
import requests
from bs4 import BeautifulSoup
import json

# -----------------------------------
# Small test set first
# -----------------------------------

test_links = [
    "https://mne.tools/stable/generated/mne.io.read_raw_edf.html",
    "https://mne.tools/stable/generated/mne.filter.filter_data.html"
]

# -----------------------------------
# Storage
# -----------------------------------

parsed_docs = []

# -----------------------------------
# Scraping Loop
# -----------------------------------

for url in mne_documentation_links:

    print(f"\nScraping: {url}")

    response = requests.get(url)

    if response.status_code != 200:
        print("Failed to fetch page")
        continue

    soup = BeautifulSoup(response.text, "html.parser")

    # -----------------------------------
    # Function Name
    # -----------------------------------

    title_tag = soup.find("h1")

    function_name = (
        title_tag.get_text(strip=True)
        if title_tag else "Unknown"
    )

    print(f"Function: {function_name}")

    # -----------------------------------
    # Description
    # -----------------------------------

    description = ""

    paragraphs = soup.find_all("p")

    if paragraphs:
        description = paragraphs[0].get_text(
            " ",
            strip=True
        )

    print(f"Description: {description[:150]}")

    # -----------------------------------
    # Parameters
    # -----------------------------------

    parameters = {}

    dl_tags = soup.find_all("dl")

    for dl in dl_tags:

        classes = dl.get("class", [])

        if "field-list" in classes:

            dt_tags = dl.find_all("dt")
            dd_tags = dl.find_all("dd")

            for dt, dd in zip(dt_tags, dd_tags):

                param_name = dt.get_text(
                    " ",
                    strip=True
                )

                param_desc = dd.get_text(
                    " ",
                    strip=True
                )

                parameters[param_name] = param_desc

    print(f"Parameters found: {len(parameters)}")

    # -----------------------------------
    # Examples
    # -----------------------------------

   # function_name = function_name.replace("#", "")

#    if description == "Section Navigation":
    #     description = ""

    # parameters.pop("Parameters :", None)
    # parameters.pop("Returns :", None)

    # examples = [
    #     ex for ex in examples
    #     if len(ex) > 20 and ">>>" not in ex
    # ]

    # -----------------------------------
    # Clean UP
    # -----------------------------------


    function_name = function_name.replace("#", "")

    description = ""

    parameters.pop("Parameters :", None)
    parameters.pop("Returns :", None)
    
    # -----------------------------------
    # Save Structured Data
    # -----------------------------------

    function_data = {
        "function_name": function_name,
        "url": url,
        "description": description,
        "parameters": parameters
    }

    parsed_docs.append(function_data)

# -----------------------------------
# Save JSON
# -----------------------------------

with open("mne_docs_test.json", "w") as f:

    json.dump(
        parsed_docs,
        f,
        indent=4
    )

print("\nSaved JSON successfully.")
```

```python
import json

# ==========================================
# Load JSON
# ==========================================

with open("mne_docs_test.json", "r") as f:
    data = json.load(f)

# ==========================================
# Convert to Semantic Text Chunks
# ==========================================

documents = []

for item in data:

    function_name = item.get(
        "function_name",
        ""
    )

    description = item.get(
        "description",
        ""
    )

    parameters = item.get(
        "parameters",
        {}
    )

    # --------------------------------------
    # Build Semantic Text
    # --------------------------------------

    text_block = f"""
Function:
{function_name}

Description:
{description}

Parameters:
"""

    for param, desc in parameters.items():

        text_block += f"""
{param}:
{desc}
"""

    documents.append({
        "function_name": function_name,
        "url": item.get("url", ""),
        "text": text_block
    })

# ==========================================
# Preview
# ==========================================

print(documents[0]["text"][:2000])
```

```python
from sentence_transformers import SentenceTransformer
import numpy as np

# ==========================================
# Load Embedding Model
# ==========================================

model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

# ==========================================
# Extract Texts
# ==========================================

texts = [
    doc["text"]
    for doc in documents
]

# ==========================================
# Generate Embeddings
# ==========================================

embeddings = model.encode(
    texts,
    show_progress_bar=True
)

# ==========================================
# Convert to NumPy
# ==========================================

embeddings = np.array(
    embeddings,
    dtype="float32"
)

print(embeddings.shape)
```

```python
import faiss
import numpy as np

# ==========================================
# Convert Embeddings to NumPy Float32
# ==========================================

embeddings = np.array(
    embeddings,
    dtype="float32"
)

# ==========================================
# Get Embedding Dimension
# ==========================================

dimension = embeddings.shape[1]

print("Embedding Dimension:", dimension)

# ==========================================
# Create FAISS Index
# ==========================================

index = faiss.IndexFlatL2(dimension)

# ==========================================
# Add Embeddings to FAISS
# ==========================================

index.add(embeddings)

# ==========================================
# Verify
# ==========================================

print(
    "Total vectors stored:",
    index.ntotal
)
```

```python
# ==========================================
# User Query
# ==========================================

query = "How can I load EDF EEG files?"

# ==========================================
# Convert Query to Embedding
# ==========================================

query_embedding = model.encode(
    [query]
)

query_embedding = np.array(
    query_embedding,
    dtype="float32"
)

# ==========================================
# Search FAISS
# ==========================================

k = 3

distances, indices = index.search(
    query_embedding,
    k
)

# ==========================================
# Display Results
# ==========================================

for rank, idx in enumerate(indices[0]):

    print("\n========================")
    print(f"Rank: {rank + 1}")
    print("========================")

    print(
        "Function:",
        documents[idx]["function_name"]
    )

    print(
        "\nRetrieved Text:\n"
    )

    print(
        documents[idx]["text"][:1500]
    )
```

```python
from sentence_transformers import CrossEncoder

# ==========================================
# Load Cross Encoder
# ==========================================

reranker = CrossEncoder(
    "cross-encoder/ms-marco-MiniLM-L-6-v2"
)

# ==========================================
# User Query
# ==========================================

query = "How can I load EDF EEG files?"

# ==========================================
# FAISS Retrieval
# ==========================================

k = 10

distances, indices = index.search(
    query_embedding,
    k
)

# ==========================================
# Prepare Query-Document Pairs
# ==========================================

pairs = []

retrieved_docs = []

for idx in indices[0]:

    text = documents[idx]["text"]

    pairs.append(
        [query, text]
    )

    retrieved_docs.append(
        documents[idx]
    )

# ==========================================
# Rerank Scores
# ==========================================

scores = reranker.predict(pairs)

# ==========================================
# Combine Results
# ==========================================

ranked_results = list(
    zip(scores, retrieved_docs)
)

# ==========================================
# Sort by Score
# ==========================================

ranked_results = sorted(
    ranked_results,
    key=lambda x: x[0],
    reverse=True
)

# ==========================================
# Display Final Results
# ==========================================

for rank, (score, doc) in enumerate(ranked_results[:3]):

    print("\n====================")
    print(f"Final Rank: {rank+1}")
    print("====================")

    print("Score:", score)

    print(
        "\nFunction:",
        doc["function_name"]
    )

    print(
        "\nRetrieved Text:\n"
    )

    print(
        doc["text"][:1200]
    )
```
